# LLMInference — Colab GPU setup
Notebook này clone repository vào Colab, kiểm tra GPU và đặt thư mục làm việc. Hãy chọn **Runtime > Change runtime type > GPU** trước khi chạy.

```python
!nvidia-smi

import torch
assert torch.cuda.is_available(), 'Chưa có GPU. Chọn Runtime > Change runtime type > GPU rồi kết nối lại.'
print('PyTorch:', torch.__version__)
print('GPU:', torch.cuda.get_device_name(0))
```

```python
# Repository public: giữ PRIVATE_REPO = False.
# Repository private: tạo Colab secret GITHUB_TOKEN và đổi thành True.
REPO_URL = 'https://github.com/kyonest12/LLMInferenceLab.git'
BRANCH = 'main'
PRIVATE_REPO = False

from pathlib import Path
from urllib.parse import urlparse
import os
import subprocess

if '<USERNAME>' in REPO_URL or '<REPOSITORY>' in REPO_URL:
    raise ValueError('Hãy thay REPO_URL bằng URL repository GitHub của bạn.')

clone_url = REPO_URL
if PRIVATE_REPO:
    from google.colab import userdata
    token = userdata.get('GITHUB_TOKEN')
    if not token:
        raise ValueError('Không tìm thấy Colab secret GITHUB_TOKEN.')
    parsed = urlparse(REPO_URL)
    clone_url = f'{parsed.scheme}://x-access-token:{token}@{parsed.netloc}{parsed.path}'

repo_dir = Path('/content/LLMInference')
if (repo_dir / '.git').exists():
    subprocess.run(['git', '-C', str(repo_dir), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', '--branch', BRANCH, clone_url, str(repo_dir)], check=True)

os.chdir(repo_dir)
print('Working directory:', Path.cwd())
```
